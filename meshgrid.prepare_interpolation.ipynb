# Generate grid description file (GDF)

### CRU - temperature

```python
tmp_cru = xr.open_mfdataset('/bettik/PROJECTS/pr-regional-climate/santolam/cru/' + 'cru_ts4.05.1901.2020.tmp.dat_HMA.nc').tmp.sel(lat=latitude,lon=longitude)
```

```python
lat2d, lon2d = np.meshgrid(tmp_cru.lat, tmp_cru.lon)
```

```python
cdlat2D = xr.DataArray(data=lat2d.transpose(),  dims=["y", "x"],
    coords=dict(x=(["x"], tmp_cru.lon.values), y=(["y"], tmp_cru.lat.values)), name='lat')
lon2D = xr.DataArray(data=lon2d.transpose(),  dims=["y", "x"],
    coords=dict(x=(["x"], tmp_cru.lon.values), y=(["y"], tmp_cru.lat.values)), name='lon')
lonlat = xr.merge([lat2D, lon2D])
lonlat.to_netcdf('grid_cru_2d_'+domain+'.nc')
```

### ERA5-land - temperature

```python
tmp_era5_land = xr.open_mfdataset('/bettik/PROJECTS/pr-regional-climate/santolam/era5-land/' + 't2m.mon_era5-land.nc').t2m
tmp_era5_land=tmp_era5_land.rename({'latitude':'lat', 'longitude':'lon'})
tmp_era5_land=tmp_era5_land.sel(lat=latitude,lon=longitude)
```
