```python
#GRf
# longitude=slice(66,78)
# latitude=slice(36,42)

#GRa
longitude=slice(81,91)
latitude=slice(25,32)
```

# Generate grid description file (GDF)

### CRU - temperature

```python
tmp_cru = xr.open_mfdataset('/bettik/PROJECTS/pr-regional-climate/santolam/cru/' + 'cru_ts4.05.1901.2020.tmp.dat.nc').tmp.sel(lat=latitude,lon=longitude)
```

```python
lat2d, lon2d = np.meshgrid(tmp_cru.lat, tmp_cru.lon)
```

```python
lat2D = xr.DataArray(data=lat2d.transpose(),  dims=["y", "x"],
    coords=dict(x=(["x"], tmp_cru.lon.values), y=(["y"], tmp_cru.lat.values)), name='lat')
lon2D = xr.DataArray(data=lon2d.transpose(),  dims=["y", "x"],
    coords=dict(x=(["x"], tmp_cru.lon.values), y=(["y"], tmp_cru.lat.values)), name='lon')
lonlat = xr.merge([lat2D, lon2D])
lonlat.to_netcdf('grid_CRU_2d_Mera.nc')
```
